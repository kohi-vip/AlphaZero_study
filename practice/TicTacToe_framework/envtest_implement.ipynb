# 🧪 Kiểm tra Môi trường & Tài nguyên Framework TicTacToe

Thư mục `practice/TicTacToe_framework/` hiện đã có đầy đủ:
- 📂 **`utils/`**: Chứa toàn bộ môi trường game (`ttt_simple_env.py`, `ttt_env.py`, `coin_env.py`, `conn_env.py`) và thuật toán nền tảng (`ch01util` -> `ch20util`).
- 📂 **`files/`**: Chứa toàn bộ trọng số mô hình đã huấn luyện trước (`.h5`) và dữ liệu tự chơi (`.p`, `.csv`).

Bây giờ bạn có thể **import trực tiếp** mà không cần qua bất kỳ cấu hình đường dẫn phức tạp nào!

```python
import os
import sys
from pathlib import Path

print("=" * 65)
print("🔍 KIỂM TRA TÀI NGUYÊN NỘI BỘ TRONG TicTacToe_framework:")
base_path = Path.cwd()
utils_path = base_path / "utils"
files_path = base_path / "files"

print(f" - Thư mục hiện tại : {base_path}")
print(f" - Thư mục utils    : {'[OK] Đã có' if utils_path.exists() else '[X] Chưa có'}")
print(f" - Thư mục files    : {'[OK] Đã có' if files_path.exists() else '[X] Chưa có'}")

# Kiểm tra số lượng file đã copy vào
if utils_path.exists():
    u_files = list(utils_path.glob("*.py"))
    print(f"   + Số file python trong utils: {len(u_files)} file")
if files_path.exists():
    f_models = list(files_path.glob("*.h5"))
    print(f"   + Số model .h5 trong files   : {len(f_models)} model")

print("\n" + "=" * 65)
print("📦 IMPORT TRỰC TIẾP MÔI TRƯỜNG TICTACTOE:")

try:
    import numpy as np
    print(f" [x] NumPy             : OK (version {np.__version__})")
except ImportError as e:
    print(f" [!] NumPy             : Chưa cài đặt ({e})")

# Import trực tiếp từ thư mục utils nội bộ
try:
    from utils.ttt_simple_env import ttt
    print(" [x] utils.ttt_simple_env : Import THÀNH CÔNG!")
    
    # Khởi tạo bàn cờ và chạy thử
    env = ttt()
    state = env.reset()
    print(f"\n🎮 Bàn cờ khởi tạo     : {state}")
    print(f" - Lượt đi đầu         : Người chơi '{env.turn}'")
    print(f" - Các ô hợp lệ        : {env.validinputs}")
    
    action = env.sample()
    print(f" - Thử nước đi ngẫu nhiên: Đánh vào ô {action}")
    env.step(action)
    print(f" - Trạng thái sau nước đi: {env.state}")
    print(f" - Các ô còn lại        : {env.validinputs}")
    print("\n===> KẾT LUẬN: Môi trường đã sẵn sàng 100% để phát triển và thử nghiệm!")
except Exception as e:
    print(f" [!] Lỗi import: {e}")

print("=" * 65)
```
